```python
parameters = {
    'trav_time_full': ['trav_time_full'],
    'trav_time_coverage': ['trav_time_coverage'],
    'l_path_full': ['l_path_full'],
    'l_path_coverage': ['l_path_coverage'],
    'yaw_path_full': ['yaw_path_full'],
    'yaw_path_coverage': ['yaw_path_coverage'],
    'area_coverage_pc': ['area_coverage_pc'],
    'area_overlap_pc': ['area_overlap_pc'],
    'pt_segmentation': ['pt_segmentation'],
    'pt_coverage_planning': ['pt_coverage_planning'],
    'pt_sequencing': ['pt_sequencing'],
    'pt_pipeline': ['pt_pipeline'],
}
```

```python
import json
import pandas as pd
from pathlib import Path
```

```python
def collect_results(parameter_path):
    """Collect all result_*.json files from Map* folders in current directory and extract relevant data."""
    data = []

    # Find all result_*.json files directly under Map* folders in current dir
    for json_file in Path('.').glob('Map*/result_*.json'):
        try:
            with open(json_file, 'r') as f:
                result = json.load(f)

            # Extract required fields
            map_name = result.get('map_name', 'Unknown')
            method = result.get('method', 'Unknown')

            # Navigate through nested dictionary using parameter_path
            value = result
            for key in parameter_path:
                value = value.get(key, {})

            # If final value is not a dict, it's our target value
            if not isinstance(value, dict) and value is not None:
                data.append({
                    'map': map_name,
                    'method': method,
                    'value': value
                })

        except (json.JSONDecodeError, FileNotFoundError, KeyError) as e:
            print(f"Error reading {json_file}: {e}")

    return data
```

```python
def create_table(data):
    """Create a pivot table with methods as rows and maps as columns."""
    if not data:
        print("No valid data found.")
        return None

    # Convert to DataFrame
    df = pd.DataFrame(data)

    # Create pivot table
    table = df.pivot_table(
        index='method',
        columns='map',
        values='value',
        aggfunc='first'  # In case of duplicates, take first value
    )

    # add rounded sum column to the table
    table['sum'] = table.sum(axis=1).round(2)

    return table
```

```python
def get_table(param, disp=True):
    parameter_path = parameters.get(param, parameters)
    data = collect_results(parameter_path)
    table = create_table(data)
    if disp:
        display(table)
    return table
```

# trav_time_full


```python
table = get_table("trav_time_full")
# best approach with minimum sum
best_approach_full_trav_time = table['sum'].idxmin()
print(f"Best approach for trav_time_full: {best_approach_full_trav_time}")
```
