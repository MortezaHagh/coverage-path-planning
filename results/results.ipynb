```python
parameters = {
    'full_trav_time': ['full_trav_time'],
    'coverage_trav_time': ['coverage_trav_time'],
    'full_path_length': ['full_path_length'],
    'coverage_path_length': ['coverage_path_length'],
    'full_yaw_rad': ['full_yaw_rad'],
    'coverage_yaw_rad': ['coverage_yaw_rad'],
    'coverage_p': ['coverage_p'],
    'overlap_p': ['overlap_p'],
    'time_segmentation': ['time_segmentation'],
    'time_coverage_planning': ['time_coverage_planning'],
    'time_sequencing': ['time_sequencing'],
    'time_total': ['time_total'],
}
```

```python
import json
import pandas as pd
from pathlib import Path
```

```python
def collect_results(parameter_path):
    """Collect all result_*.json files and extract relevant data."""
    data = []

    # Find all result_*.json files recursively
    for json_file in Path('..').rglob('result_*.json'):
        try:
            with open(json_file, 'r') as f:
                result = json.load(f)

            # Extract required fields
            map_name = result.get('map_name', 'Unknown')
            method = result.get('method', 'Unknown')

            # Navigate through nested dictionary using parameter_path
            value = result
            for key in parameter_path:
                value = value.get(key, {})

            # If final value is not a dict, it's our target value
            if not isinstance(value, dict) and value is not None:
                data.append({
                    'map': map_name,
                    'method': method,
                    'value': value
                })

        except (json.JSONDecodeError, FileNotFoundError, KeyError) as e:
            print(f"Error reading {json_file}: {e}")

    return data
```

```python
def create_table(data):
    """Create a pivot table with methods as rows and maps as columns."""
    if not data:
        print("No valid data found.")
        return None

    # Convert to DataFrame
    df = pd.DataFrame(data)

    # Create pivot table
    table = df.pivot_table(
        index='method',
        columns='map',
        values='value',
        aggfunc='first'  # In case of duplicates, take first value
    )

    # add rounded sum column to the table
    table['sum'] = table.sum(axis=1).round(2)

    return table
```

```python
def get_table(param, disp=True):
    parameter_path = parameters.get(param, parameters)
    data = collect_results(parameter_path)
    table = create_table(data)
    if disp:
        display(table)
    return table
```

# full_trav_time


```python
table = get_table("full_trav_time")
# best approach with minimum sum
best_approach_full_trav_time = table['sum'].idxmin()
print(f"Best approach for full_trav_time: {best_approach_full_trav_time}")
```
